```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Load and subsample the data
sample_size = 10000  # Adjust this based on your needs
np.random.seed(42)  # For reproducible results

# Subsample the data
idx_test = np.random.choice(len(hfile_r['test/data/data']), 
                           min(sample_size, len(hfile_r['test/data/data'])), 
                           replace=False)
idx_exp = np.random.choice(len(hfile_e['exp/raw/data/part_s2020_c01_r0027/data']), 
                          min(sample_size, len(hfile_e['exp/raw/data/part_s2020_c01_r0027/data'])), 
                          replace=False)

data_test = hfile_r['test/data/data'][idx_test]
data_exp = hfile_e['exp/raw/data/part_s2020_c01_r0027/data'][idx_exp]

print(f"Subsampled data shapes: Test {data_test.shape}, Exp {data_exp.shape}")

# Feature names
features = ['Q', 't', 'x', 'y', 'z']

# Set up the plotting style
plt.style.use('default')

# Create main comparison plot
fig, axes = plt.subplots(2, 5, figsize=(20, 10))
fig.suptitle(f'Feature Comparison (n={sample_size} samples each)', fontsize=16)

for i, feature in enumerate(features):
    # Row 1: Overlaid histograms
    axes[0, i].hist(data_test[:, i], bins=30, alpha=0.6, label='Test', 
                    density=True, color='blue', edgecolor='black', linewidth=0.5)
    axes[0, i].hist(data_exp[:, i], bins=30, alpha=0.6, label='Exp', 
                    density=True, color='red', edgecolor='black', linewidth=0.5)
    axes[0, i].set_title(f'{feature} - Distribution')
    axes[0, i].set_xlabel(feature)
    axes[0, i].set_ylabel('Density')
    axes[0, i].legend()
    axes[0, i].grid(True, alpha=0.3)
    
    # Row 2: Box plots
    bp = axes[1, i].boxplot([data_test[:, i], data_exp[:, i]], 
                           labels=['Test', 'Exp'], patch_artist=True)
    bp['boxes'][0].set_facecolor('lightblue')
    bp['boxes'][1].set_facecolor('lightcoral')
    axes[1, i].set_title(f'{feature} - Box Plot')
    axes[1, i].set_ylabel(feature)
    axes[1, i].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Quick statistical comparison
print("\nQuick Statistical Summary:")
print("="*70)
print(f"{'Feature':<8} {'Dataset':<8} {'Mean':<10} {'Std':<10} {'Min':<10} {'Max':<10}")
print("-"*70)

for i, feature in enumerate(features):
    for data, name in [(data_test, 'Test'), (data_exp, 'Exp')]:
        stats_vals = [np.mean(data[:, i]), np.std(data[:, i]), 
                     np.min(data[:, i]), np.max(data[:, i])]
        print(f"{feature if name=='Test' else '':<8} {name:<8} {stats_vals[0]:<10.3f} {stats_vals[1]:<10.3f} {stats_vals[2]:<10.3f} {stats_vals[3]:<10.3f}")

# Simple correlation comparison
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

for ax, data, title in [(ax1, data_test, 'Test Data'), (ax2, data_exp, 'Exp Data')]:
    corr = np.corrcoef(data.T)
    im = ax.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
    ax.set_title(f'{title} - Correlation Matrix')
    ax.set_xticks(range(5))
    ax.set_yticks(range(5))
    ax.set_xticklabels(features)
    ax.set_yticklabels(features)
    
    # Add correlation values
    for i in range(5):
        for j in range(5):
            ax.text(j, i, f'{corr[i, j]:.2f}', ha='center', va='center',
                   color='white' if abs(corr[i, j]) > 0.5 else 'black', fontsize=10)

plt.colorbar(im, ax=[ax1, ax2], shrink=0.8)
plt.tight_layout()
plt.show()

# Key spatial relationships (subsample further for scatter)
scatter_sample = 2000
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

scatter_pairs = [('x', 'y', 2, 3), ('x', 'z', 2, 4), ('y', 'z', 3, 4)]

for idx, (feat1, feat2, i1, i2) in enumerate(scatter_pairs):
    # Further subsample for scatter plots
    sample_test_scatter = data_test[:scatter_sample]
    sample_exp_scatter = data_exp[:scatter_sample]
    
    axes[idx].scatter(sample_test_scatter[:, i1], sample_test_scatter[:, i2], 
                     alpha=0.6, s=3, label='Test', color='blue')
    axes[idx].scatter(sample_exp_scatter[:, i1], sample_exp_scatter[:, i2], 
                     alpha=0.6, s=3, label='Exp', color='red')
    axes[idx].set_xlabel(feat1)
    axes[idx].set_ylabel(feat2)
    axes[idx].set_title(f'{feat1} vs {feat2}')
    axes[idx].legend()
    axes[idx].grid(True, alpha=0.3)

plt.suptitle(f'Spatial Relationships (n={scatter_sample} samples)', fontsize=14)
plt.tight_layout()
plt.show()

print(f"\nOriginal data sizes:")
print(f"Test: {hfile_r['test/data/data'].shape[0]:,} samples")
print(f"Exp: {hfile_e['exp/raw/data/part_s2020_c01_r0027/data'].shape[0]:,} samples")
print(f"Analyzed: {sample_size:,} samples each")
```
